```python
import json
import pandas as pd
import matplotlib.pyplot as plt

file_path = 'raw_results.json'
with open(file_path, 'r') as f:
    data = json.load(f)

raw_values = data[0]['primaryMetric']['rawData'][0]
iterations = list(range(1, len(raw_values) + 1))

plt.figure(figsize=(10, 6))
plt.plot(iterations, raw_values, marker='o', linestyle='-', color='b', label='Benchmark')

plt.yscale('log')

plt.title('JMH Benchmark Verlauf (Log-Skala) with warmup')
plt.xlabel('Iteration #')
plt.ylabel('Time (ms/op)')
plt.grid(True, which="both", ls="-", alpha=0.5)
plt.legend()
plt.show()
```
